```python
import pandas as pd
import numpy as np

# 1. LOAD THE GROUND TRUTH DATA

# Manually load the different generated data sets for both the cases

print("Loading ground truth dataset for Patterns C...")
df_truth = pd.read_csv('MICE_wide_3jamnew.csv')

# Ensure Timestamp is a datetime object
df_truth['Timestamp'] = pd.to_datetime(df_truth['Timestamp'])


# PATTERN C: THE ATYPICAL SHOCK
# Deleting S4 ONLY in dir_a (where the jam actually happened)

df_pattern_c = df_truth.copy()

#not deleteting the entire traffic jam, deleteing only data between after 30 minutes and before 30 minutes of the congestion.
jam_start = pd.to_datetime('2026-07-25 11:30:00')
jam_end = pd.to_datetime('2026-07-25 12:45:00')

jam_mask = (df_pattern_c['Timestamp'] >= jam_start) & (df_pattern_c['Timestamp'] <= jam_end)

# Target ONLY dir_a for the shockwave deletion
shockwave_cols = ['Spd_S4_dir_a', 'Vol_S4_dir_a']
df_pattern_c.loc[jam_mask, shockwave_cols] = np.nan

```

```python
# Export the new curated dataset
# change the csv file for each of the data generated.
df_pattern_c.to_csv('traffic_pattern_c_wide_3jamnew.csv', index=False)
print("Pattern C (Atypical Shock - dir_a only) saved.")
```
